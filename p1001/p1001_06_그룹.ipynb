```python
# 그룹핑 
df.groupby('학교').count()
df['학교'].value_counts() # df.groupby('학교').size()와 동일한 기능 
```

```python
df['학년'] = [3,3,2,1,1,3,2,1] # 학년 추가 
df.groupby("학교")['학년'].value_counts().loc['구로고'] # row에 대한 정보:loc 
df.groupby("학교")['학년'].value_counts().loc['디지털고']
df.groupby("학교")['학년'].value_counts()

# 백분율로 표시 normalize : 원형 그래프 만들기 가능 
df.groupby("학교")['학년'].value_counts(normalize=True).loc['구로고'] 
```

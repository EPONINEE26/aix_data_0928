```python
import pandas as pd
df = pd.read_csv('file/chipotle.tsv', sep='\t') # tsv 읽어올 때 sep='\t'
# df.describe() # 기본 통계
# df.info() # 기본 정보 
# df.columns # 컬럼 정보 
# df['order_id'].unique() # 주문서 숫자
# df['order_id'] = df['order_id'].astype(int)
# df['item_price'] = df['item_price'].str.strip().replace('$', '').astype(float) # 미리 계산하는 것도 좋음 
```

```python
# 1. item_name 당 주문 개수를 출력하시오. 10개만 출력
df['item_name'].value_counts().head(10)
```

```python
# 2. item_name 당 주문 총량(quantity)을 출력하시오.
df['quantity'].value_counts().head(10)
```

```python
# 3. 가장 많이 주문한 item_name  10개를 출력하시오.
df.groupby('item_name')['quantity'].max().sort_values(ascending=False).head(10)
```

```python
# 4. 주문당 평균 금액을 출력하시오.
df['item_price'] = (df['item_price'].astype(str).str.replace('$', '').astype(float))
df.groupby('item_name')['item_price'].sum().mean()
```

```python
# 5. 한 주문에 10달러 이상 사용한 주문 번호(id) 출력하시오.
df.groupby('order_id')['item_price'].sum().mean()

flit = df.groupby('order_id')['item_price'].sum()>10
df.groupby('order_id')['item_price'].sum()[flit]
```

```python
# Chicken Bowl을 2개 이상 구매한 주문서 출력하시오.

flit = (df['item_name'] == 'Chicken Bowl') & (df['quantity'] >=2)
df[flit]
df[flit][['item_price','order_id']]
```

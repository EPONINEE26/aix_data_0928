```python
import pandas as pd
df=pd.read_excel('file/students.xlsx', index_col='번호')

# df
```

```python
# 합계 컬럼 추가
# SW특기 있으면 합계점수 5점 추가해서 입력하시오 
# 함수를 적용해서 처리하시오 

df['합계'] = df['국어'] + df['영어'] + df['수학'] 

def capChange(SW특기):
    if SW특기: 
        return 5
    else:
        return 0

df['가산점'] = df['SW특기'].apply(capChange)
df['합계'] = df['국어'] + df['영어'] + df['수학'] + df['가산점']
df


# null 비교 
# df['합계'] = df['국어'] + df['영어'] + df['수학']

# def cal_sw(lang):
#     if pd.notnull(lang): # null이 아니면 
#     # if lang != '':  
#         return lang.str.lower()
#     else:
#         return "없음"
# df['SW특기'] = df['SW특기'].apply(cal_sw)
# df 
```

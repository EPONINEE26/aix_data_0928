```python
import pandas as pd
data = {
    '영화' : ['명량', '극한직업', '신과함께-죄와 벌', '국제시장', '괴물', '도둑들', '7번방의 선물', '암살'],
    '개봉 연도' : [2014, 2019, 2017, 2014, 2006, 2012, 2013, 2015],
    '관객 수' : [1761, 1626, 1441, 1426, 1301, 1298, 1281, 1270], # (단위 : 만 명)
    '평점' : [8.88, 9.20, 8.73, 9.16, 8.62, 7.64, 8.83, 9.10]
}
df = pd.DataFrame(data)
# df.info()

```

```python
# 1. 함수사용 - 천단위 표시
def cal_change(x):
    return  f"{x:,}"
    # return  '{:,d}'.format() # 숫자를 문자변형

df['관객 수'] = df['관객 수'].apply(cal_change)

#  2. lambda함수적용
df['관객 수'] = df['관객 수'].apply(lambda x: f"{x:,}")
#  3. map사용
# df['관객 수'] = df['관객 수'].map('{:,}'.format)
```

```python
# 관객수 1761만명 출력 / 천단위 출력 '{:d}'.format() # 함수사용 
# 관객수 1,761만명 출력 
df['관객 수'] = df['관객 수'].astype(int)
df['관객 수']=df['관객 수']+'만명'
df['관객 수'] = df['관객 수'].astype(str) + '만명'
```

```python
# map 함수 
df['관객 수']=df['관객 수'].map('{:,}'.format) # 함수 한 개씩 가져와서 처리해줌 
# lambda 함수 
df['관객 수']=df['관객 수'].apply(lambda x: f"{x:},")
df

def cal(x):
    return x+2

#(lambda x : x+2) # 1줄 함수 
df
```

```python
import pandas as pd
df = pd.read_excel('file/students.xlsx', index_col='번호')
df
# colds 리스트 순서를 변경 
# 1. cols은 리스트 타입으로 변환 
cols = list(df.columns) # cols은 리스트 타입으로 변환 

# 2. 
cols[-2:] # 끝에서 두번째부터 출력 (평균,결과 컬럼)
cols[:-2] # 처음부터 끝에서 두번째 전까지 출력 (이름 ~ 합계)
cols[0]
cols[1:-2] # 1번부터 끝에서 두번째 전까지 출력 (학교 ~ 수학) # list 타입 
type(cols[0]) # str # 리스트 타입일 때만 덧셈 가능 
type(cols[1:-2]) # list 

# 3. 리스트 순서변경 
[cols[0]]+cols[-2:]+cols[1:-2] # str+#list+list => error / str에 대괄호 추가하면 가능 
# # 변경된 리스트 적용
df=df[cols]

```

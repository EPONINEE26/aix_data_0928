```python
import pandas as pd
df = pd.read_excel('file/score.xlsx', index_col='지원번호')

# 1. 합계 (과목총점을 계산해서) 컬럼을 추가하시오
# 2. 평균 컬럼을 추가하시오
# 3.등급 컬럼을 추가해서 90점 이상 A, 80점이상 B,70점 이상 C,60점 이상 D, 그 외에는 F를 입력하시오
# 함수를 사용해서 처리하시오

df['합계'] = 0
df['합계'] = df['국어']+df['영어']+df['수학']+df['과학']+df['사회']
df['평균'] = df['합계'] / 5
df['등급'] = 0 

# df['score'] = df['평균']
def capChange(score):
    if score > 90: return 'A' 
    elif score > 80: return 'B' 
    elif score > 70: return 'C' 
    elif score > 60: return 'D' 
    else : return 'F'
df['등급'] = df['score'].apply(capChange)
df

```

```python
# 합계에 등수를 만들어 줌 (ascending=False : 역순정렬 method='min' : 1,2,2,4로 출력. 같은 )
df['등수'] = df['합계'].rank(ascending=False).astype(int) # 합계에 등수를 만들어 줌
df
```

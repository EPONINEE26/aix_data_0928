```python
df = pd.read_csv('file/namu2.csv')
df = df.head(5)
```

```python
def text_cleaning(text):
    if pd.isna(text):
        return ''
    hangul = re.compile('[^ ㄱ-ㅣ가-힣]+')    # 정규표현식에 해당이 되는지 확인
    return hangul.sub('',text)
```

```python
df['content'] = df['content'].apply(lambda x:text_cleaning(x))
df['content']
df_content_all ="".join(df['content']) # 리스트에 있는 것을 1개의 문자열로 합침 
df_content_all 

```

```python
n_tag = Okt()
nouns = n_tag.nouns(df['content'][0])
print(nouns)

count = Counter(nouns) 
count # 딕셔너리 타입 
```

```python
# 1개 단어 키워드 제외
# count : 딕셔너리 타입 
# remove_counter = Counter({'당신' : 2}) # '당신' : 2 이런식으로 출력해라... 의미 

# for x in count:
#     if len(x)>1:
#         {x:count[x]}
        
remove_counter = Counter({x:count[x] for x in count if len(x)>1}) # 1보다 큰 글자만 x로 넣어라 그 이후 remove_counter에 넣어라. 위와 동일한 의미 
remove_counter # 한 글자가 삭제됨 
```

```python
df['total_servings'] = df['beer_servings'] + df['spirit_servings'] + df['wine_servings']
df['total_servings']
```

```python
df.shape
```

```python
idx = np.arange(df.shape[0])
idx
```

```python
# 한국 막대그래프를 red 추가해서 출력하시ㅗㅇ.
sort_df = df.sort_values('total_servings',ascending=False)
sort_df['country']
sort_df['total_servings']
k_sort_df = sort_df[sort_df['country']=='South Korea']
x = sort_df['country']        #전체 국가이름
y = sort_df['total_servings'] #전체 국가별 합계
x2 = k_sort_df['country']  # 한국이름
y2 = k_sort_df['total_servings'] #한국합계

plt.bar(x,y,label='국가별')
plt.bar(x2,y2,color='r',label='한국') #한국그래프
plt.legend()
plt.xticks(x,idx)
plt.show()
```

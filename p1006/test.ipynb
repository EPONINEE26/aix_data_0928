```python
from konlpy.tag import Okt 
from collections import Counter

# konlpy : 미래형이나 과거형은 현재형으로 변환하여 출력하는 기능
text = '나는 당신을 사랑합니다. 감사합니다. 고맙습니다. 당신이 최고입니다.'
n_tag = Okt()
nouns = n_tag.nouns(text)
print(nouns) # 명사만 출력하라는 의미 

count = Counter(nouns)
count
```

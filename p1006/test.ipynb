```python
from konlpy.tag import Okt 
from collections import Counter

# konlpy : 미래형이나 과거형은 현재형으로 변환하여 출력하는 기능
text = '나는 당신을 사랑합니다. 감사합니다. 고맙습니다. 당신이 최고입니다.'
n_tag = Okt()
nouns = n_tag.nouns(text)
print(nouns) # 명사만 출력하라는 의미 

count = Counter(nouns)
count
```

```python
from selenium import webdriver
from selenium.webdriver.common.by import By
from selenium.webdriver.common.keys import Keys
from selenium.webdriver.chrome.options import Options
import requests
from bs4 import BeautifulSoup
import time
import os
from dotenv import load_dotenv
```

```python
url = "https://namu.wiki/RecentChanges"
options = Options()
options.add_experimental_option("excludeSwitches", ["enable-automation"])
options.add_experimental_option("useAutomationExtension", False)
options.add_argument("--disable-blink-features=AutomationControlled")
browser = webdriver.Chrome(options=options)
browser.maximize_window() # 화면 최대창 확대
browser.get(url)
time.sleep(2)
# # 파일저장
soup = BeautifulSoup(browser.page_source,'lxml')
with open('file/namu1.html','w',encoding='utf-8') as f:
    f.write(soup.prettify())
```

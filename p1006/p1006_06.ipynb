```python
namu_data=[]
for div in divs :
    title_tag = div.find('a')
    title = title_tag.get_text(strip=True) if title_tag else None
    # print(title_tag)
    first_url = 'https://namu.wiki'
    url = first_url+title_tag['href']
    # print(url)
    namu_data.append({
        'title': title,
        'url': url
    })

# namu_data = []
# for div in divs:
#     # 문서명
#     title_tag = div.find('a')
#     title = title_tag.get_text(strip=True) if title_tag else None
#     # print(title)
#     first_url = 'https://namu.wiki'
#     # url = title_tag['href']
#     url = first_url+title_tag['href']
#     # print(url)
#     namu_data.append({
#         'title':title,
#         'url':url
#     })
    
# namu_data
```

```python
data = []
data.append({
    'total' : 'SNH48Group',
    'url' : 'https://namu.wiki'
})

# 2. 다른 학생들이 F12로 힘들게 찾은 tr, td 내부 글자들을 판다스가 알아서 다 가져옵니다.
url = "https://namu.wiki"
df_list = pd.read_html(url, storage_options={'User-Agent': 'Mozilla/5.0'})

# 3. 긁어온 표에서 글자(total)와 주소(url)를 사용자님 상자에 딱 100개만 자동으로 쌓아줍니다.
for item in df_list[0].values.flatten()[:100]:
    if pd.notna(item) and type(item) == str:
        if item not in ['접기', '보기', '편집', '토론', '역사']:
            data.append({
                'total': item,
                'url': 'https://namu.wiki'
            })

# 4. 최종 데이터프레임 표 완성
df = pd.DataFrame(data)

# 5. 수업 시간 규칙대로 내 컴퓨터 폴더 안에 진짜 엑셀 파일로 바로 저장합니다.
df.to_excel('SNH48_Perfect_100.xlsx', index=False)

# 화면에 표 출력하기
df
```

```python
import matplotlib.pyplot as plt
import matplotlib
matplotlib.rcParams['axes.unicode_minus'] = False 
matplotlib.rcParams['font.family'] = 'Malgun Gothic' 
matplotlib.rcParams['font.size'] 
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv('file/chipotle.tsv', sep='\t') # tsv 불러올 때 sep='\t' 써야함
df.head(3)
```

```python
# item_name으로 그룹핑해서 quantity 합계를 가지고 막대그래프를 출력하시오.

df.groupby('item_name').count().head(3)
item_quantity = df.groupby('item_name')['quantity'].sum().head(3)

s_sum = df.groupby('item_name')['quantity'].sum().head(3)
# display(s_sum) 
s_sum
```

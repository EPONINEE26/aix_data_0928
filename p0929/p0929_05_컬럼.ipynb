```python
import pandas as pd
```

```python
# skiprow : 상단부분 제외하고 불러오기 
df = pd.read_excel('file/j2025.xlsx',skiprows=2,index_col='행정기관')
df
```

```python
# : concat : DataFrame 합치기 
df1 = pd.read_csv('file/2014년졸음운전교통사고.csv',encoding='euc-kr')
df2 = pd.read_csv('file/2015년졸음운전교통사고.csv',encoding='euc-kr')
df3 = pd.read_csv('file/2016년졸음운전교통사고.csv',encoding='euc-kr')
# df3
df = pd.concat([df1, df2, df3]) # DataFrame 합치기 
df.set_index('구분', inplace=True)
df 
```

```python
# usecols : 컬럼 선택해서 가져오기 
# df = pd.read_excel('file/j2025.xlsx',skiprows=2,usecols='B,D') # b, d 칸만 불러옴 
df = pd.read_excel('file/j2025.xlsx',skiprows=2,usecols='B:D, I:N') # b, d, i~n 칸만 불러옴 
df
```

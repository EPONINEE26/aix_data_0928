# 데이터 추가 : 없는 컬럼값을 입력하면 추가가 됨 

```python
import pandas as pd
df = pd.read_excel('file/score.xlsx', index_col='지원번호')
```

```python
# 데이터 추가 : 없는 컬럼값 입력하면 추가 가능 

df['합계'] = df['국어']+df['영어']+df['수학']+df['과학']+df['사회']
df['결과'] = '불합격' # 데이터 추가 

df
```

```python

# 185 이상 출력하시오.
# df[df['키']>=185]
# 350점 이상

# 컬럼 전체를 수정할때 df['컬럼']
# df[df['합계']>=350]['결과'] = '합격'  
# df['국어'] = df['국어']+10
# 컬럼의 부분적인 값을 넣을때는 loc,iloc를 사용
df.loc[df['합계']>=350,'결과'] = '합격'  
df 
```

```python
import pandas as pd
df = pd.read_excel('file/score.xlsx', index_col='지원번호')

# index 정렬 (순차정렬 True, 역순정렬 False)
df.sort_index(ascending=False)
df.sort_values('키', ascending=False) # 각각의 컬럼으로 정렬 가능 

df.sort_values(['국어','영어'], ascending=False) # 동점일 경우 국어가 아닌 뒤에 있는 영어로 역순정렬 

df.sort_values(['국어','영어'], ascending=[False, True]) # 국어점수는 역순, 영어점수는 순차로 할 때 

```

```python
df = pd.read_csv('file/jumin08_2.csv',index_col='행정구역')
df2 = df[['2026년08월_총인구수','2026년08월_세대수']]
df2['2026년08월_총인구수'] = df2['2026년08월_총인구수'].str.replace(',','').astype(int)
df2['2026년08월_세대수'] = df2['2026년08월_세대수'].str.replace(',','').astype(int)
df2
```

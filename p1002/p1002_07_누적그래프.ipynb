```python
import matplotlib.pyplot as plt
import matplotlib
matplotlib.rcParams['axes.unicode_minus'] = False 
matplotlib.rcParams['font.family'] = 'Malgun Gothic' 
matplotlib.rcParams['font.size'] 
import pandas as pd
import numpy as np
```

```python
df=pd.read_excel('file/score.xlsx', index_col='지원번호')
df
```

```python
# x = df.index
x = df['이름']
x = df['이름']
y = df['국어']
y2 = df['영어']
y3 = df['수학']

plt.figure(figsize=(10,5))
colors =['red','green','blue']
plt.title('학생성적프로그램')
plt.bar(x, y, label='국어', width=0.5)
plt.bar(x, y2,bottom=y, label='영어', width=0.5)
plt.bar(x, y3,bottom=y+y2, label='수학', width=0.5)

# ncols = 3 : 범례가 1줄로 나타남.
plt.legend(ncol=3)
plt.grid(axis='y', linestyle=':', alpha=0.7)

# zip 으로 묶어서 한꺼번에 가져올 수 있음 
for i,(v,v2,v3) in enumerate(zip(y,y2,y3)):
    plt.text(i,v/2,v, ha='center') # 1개 값 입력 
    plt.text(i,v+v2/2,v2, ha='center', va='center') 
    plt.text(i,v+v2+v3/2,v3, ha='center', va='center') 

# for i,v2 in enumerate(y2):
#     plt.text(i,v+v2-2,v2, ha='center')  
    
# for i,v3 in enumerate(y3):
#     plt.text(i,v+v3,v3, ha='center')  

plt.show()
```

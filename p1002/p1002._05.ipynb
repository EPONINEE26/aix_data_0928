```python
import matplotlib.pyplot as plt
import matplotlib
matplotlib.rcParams['axes.unicode_minus'] = False 
matplotlib.rcParams['font.family'] = 'Malgun Gothic' 
matplotlib.rcParams['font.size'] 
import pandas as pd
import numpy as np
```

```python
x = [1,2,3]
y = [2,4,8]

plt.title('꺾은 선')

# figure : 그래프 사이즈 변경(가로, 세로), figsize : 크기,  dpi : 해상도
# plt.figure(figsize=(20,5)), facecolor: 배경색 변경, 
# savefig : 저장 기능 
# plt.figure(figsize=(15,2), dpi=200, facecolor='yellow')
# plt.figure(figsize=(15,2), dpi=200, facecolor='yellow')
# plt.figure(figsize=(15,2), facecolor='yellow')

plt.plot(x,y,linewidth=2,marker='o',ms=10,mfc='yellow',linestyle='-.',color='blue')

# 그래프 안에 글자를 입력 : plt.text, ha:글자 중앙에 넣는다는 의미 
# plt.text(x[0],y[0]+0.3,2, ha='center', color='red') # 0.3 만큼 올린다는 의미 

for idx,value in enumerate(y):
    # plt.text(x[idx],y[idx]+0.2,value, ha='center', color='red') # 위쪽 
    plt.text(x[idx]-0.05,y[idx]-0.2,value, ha='center', color='red') # 아래쪽 
    
# X선 설명부분
# plt.xlabel('매출액',color='red',loc='right') 
# plt.ylabel('월별',color='#00aa00',loc='top') 


# 그래프 간격조정 ( 입력값 형태로 구성 )
# plt.xticks([1,1.2,1.4,1.6,2,3])
# plt.yticks([-15,10,20,30]) 
# plt.savefig('file/aaa2.png', dpi=200) # savefig : 스크린 샷 저장 가능 
plt.show()
```

```python
# 선 스타일
x = [1,2,3]
y = [20,40,80]
y2 = [50,10,30]
y3 = [10,20,50]

plt.title('꺾은선 graph')  # 상단타이틀
# plt.figure(figsize=(10,5),facecolor='yellow')
plt.plot(x,y,label='국어', linewidth=2,marker='o',ms=10,mfc='yellow',linestyle='-',\
         color='pink')
plt.plot(x,y2,label='영어',linewidth=2,marker='x',ms=10,mfc='yellow',linestyle=':',\
         color='blue')
plt.plot(x,y3,label='수학',linewidth=2,marker='^',ms=10,mfc='yellow',linestyle='-.',\
         color='green')
plt.legend()
# 그래프 안에 글자를 입력 : plt.text
for idx,value in enumerate(y): # 0,1,2
    plt.text(x[idx],y[idx]+2,value,ha='center',color='red')
for idx,value in enumerate(y2): # 0,1,2
    plt.text(x[idx],y2[idx]+2,value,ha='center',color='red')
for idx,value in enumerate(y3): # 0,1,2
    plt.text(x[idx],y3[idx]+2,value,ha='center',color='red')


# 그래프 간격조정 ( 입력값 형태로 구성 )
# plt.xticks([1,1.2,1.4,1.6,2,3])
# plt.yticks([-15,10,20,30,100]) #간격부분 조정을 하면 꺾이는 상태를 조정
plt.show()
```

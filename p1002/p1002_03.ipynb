```python
# 주문서에서 가장 많이 주문된 메뉴는 무엇인지 출력하시오. 

# df.groupby(['item_name']).count().sort_values('order_id',ascending=False).head(10)
df['item_name'].value_counts().head(10)
# df['item_name'].unique()
```

```python
# 가장 주문이 많이 된 메뉴의 총 개수 
# 주문서 2개, 3개 등등의 정보 -> 총 몇개가 주문이 들어왔는지 출력하시오
df.groupby(['item_name'])['quantity'].sum().sort_values(ascending=False).head(10)
df
```

```python
# 막대그래프로 출력하시오 
item_graph = df.groupby(['item_name'])['quantity'].sum().head(50)
x = item_graph.index
y = item_graph.values

plt.title('메뉴별 주문수량')
plt.bar(x,y)
plt.show()
```

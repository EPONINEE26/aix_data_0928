```python
import matplotlib.pyplot as plt
import matplotlib
matplotlib.rcParams['axes.unicode_minus'] = False 
matplotlib.rcParams['font.family'] = 'Malgun Gothic' 
matplotlib.rcParams['font.size'] 
import pandas as pd
import numpy as np
```

```python
df=pd.read_excel('file/score.xlsx', index_col='지원번호')
df
```

```python
# index=[1,2,3,4,5,6,7,8]
index=np.arange(df.shape[0])
x = df['이름']
x = df['이름']
y = df['국어']
y2 = df['영어']
y3 = df['수학']

plt.figure(figsize=(10,5))
colors =['red','green','blue']
plt.title('학생성적프로그램')
plt.bar(index-0.2,y,label='국어',width=0.2)
plt.bar(index,y2,label='영어',width=0.2)
plt.bar(index+0.2,y3,label='수학',width=0.2)

# 바 위에 성적을 출력하시오.
for i,(v,v2,v3) in enumerate(zip(y,y2,y3)):
    plt.text(i-0.2,v+1,v,ha='center')
    plt.text(i,v2+1,v2,ha='center')
    plt.text(i+0.2,v3+1,v3,ha='center')



plt.legend(ncol=3)
plt.xticks(index,df['이름']) # index의 위치에 df['이름']으로 대체 
plt.yticks([0, 50, 60, 70, 80, 90, 100, 120])

plt.show()
```

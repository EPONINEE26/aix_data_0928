```python
result = df.groupby(['학교', '학년'])[['국어', '영어', '수학']].mean().round(2)
result
```

```python
# students.xlsx 파일을 읽어와서 학교별, 학년별, 국어, 영어, 수학 평균을 출력하시오.

# 이 코드는 판다스 데이터프레임의 인덱스(표의 맨 왼쪽 기준선)가 2단 이상으로 복잡하게 겹쳐져 있을 때, 이걸 평평하게 풀어주는(정리하는) 코드
# "지금 result 데이터의 인덱스가 여러 겹으로 복잡하게 꼬인 멀티 인덱스(MultiIndex) 상태인지 확인해라" 라는 뜻/ (보통 groupby나 피벗 테이블 등을 쓸 때 인덱스가 2줄 이상으로 층층이 쌓이게 될 때가 많음)
# MultiIndex인 경우에만 reset_index 실행 (중복 에러 방지)

if isinstance(result.index, pd.MultiIndex):
# 만약 인덱스가 복잡하게 겹쳐 있다면, 그 층층이 쌓인 인덱스들을 일반적인 데이터 열(Column)로 쑥 내려서 표를 평평하게 만들어라는 뜻    
    result = result.reset_index()

# 학교별 평균, 학년별 평균 
df_school = result.groupby('학교')[['국어', '영어', '수학']].mean()
df_grade = result.groupby('학년')[['국어', '영어', '수학']].mean()

# 학년 인덱스에 '학년' 글자 강제로 붙이기
df_grade.index = [str(g) if '학년' in str(g) else f"{g}학년" for g in df_grade.index]

# 학교명과 학년명이 모두 x축에 나오도록 합치기
df_combined = pd.concat([df_school, df_grade])

x = np.arange(len(df_combined))  # 총 6개 (학교 3개 + 학년 3개)
width = 0.25  # 막대 두께 키우기

# 과목별 데이터 추출
kor = df_combined['국어'].values
eng = df_combined['영어'].values
math = df_combined['수학'].values

labels = df_combined.index.tolist()  # ['구로고', '신림고', '디지털고', '1학년', '2학년', '3학년']

plt.figure(figsize=(12, 6))
plt.title('학교별 및 학년별 과목 성적 그래프')

# x축에는 학교명/학년명이 오고, 바(Bar)는 국어, 영어, 수학으로 출력
plt.bar(x - width, kor, label='국어', width=width)
plt.bar(x, eng, label='영어', width=width)
plt.bar(x + width, math, label='수학', width=width)

plt.xticks(x, labels)
plt.yticks([0, 50, 60, 70, 80, 90, 100])

# 각 막대 위에 값 표시
for i in range(len(x)):
    plt.text(i - width, kor[i] + 1, f'{kor[i]:.1f}', ha='center', fontsize=7)
    plt.text(i, eng[i] + 1, f'{eng[i]:.1f}', ha='center', fontsize=7)
    plt.text(i + width, math[i] + 1, f'{math[i]:.1f}', ha='center', fontsize=7)

plt.legend(ncol=3)
plt.tight_layout() # 그래프 요소들이 서로 겹치지 않도록 여백을 자동으로 조절해주는 함수
plt.show()

# pandas로 출력 (print 사용 안 함)
display(df_school)
display(df_grade)

```
